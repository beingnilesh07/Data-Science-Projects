1] Program to find  Second Largest Number

```python
from typing import List

def getSecondOrderElement(n:int, a:List[int]) -> List[int]:
  small=float('inf')
  second_small=float('inf')
  large=float('-inf')
  second_largest=float('-inf')

  for i in range(n):
    if a[i] < small :
      second_small=small
      small=a[i]
    elif a[i] < second_small and a[i] != small:
      second_small=a[i]

    if a[i] > large:
      second_largest=large
      large=a[i]
    elif a[i] > second_largest and a[i] != large:
      second_largest = a[i]

  return [second_small, second_largest]

if __name__ == "__main__":
  arr=[24, 53, 66, 69, 33, 22]
  n=len(arr)
  result=getSecondOrderElement(n, arr)
  print('Second smallest:', result[0])
  print('Second largest:', result[1])
```

2] Find missing number in an array

```python
#Brute Search
nums = [9, 6, 4, 2, 3, 5, 7, 0, 1]

n=len(nums)
for i in range(0, n+1):
  if i not in nums:
    print(i)

# TC - O(n2)
# sc- O(1)
```

```python
#Better search

nums = [9, 6, 4, 2, 3, 5, 7, 0, 1]

n=len(nums)
freq={}
for i in range(0, n+1):
  freq[i]=0

for num in nums:
  freq[num]=1

for k, v in freq.items():
  if v == 0:
    print(k)
```

3] Find the maximum subarray sum

```python
def find_max_subarray_sum(nums):
  n=len(nums)
  maxi=float('-inf')
  start_idx = 0
  end_idx = 0

  for i in range(0, n):
    total=0
    for j in range(i, n):
      total=total + nums[j]
      if total > maxi:
        maxi = total
        start_idx = i
        end_idx = j

  max_subarray = nums[start_idx : end_idx + 1]
  return maxi, max_subarray

# Example usage:
nums = [-2, 1, -3, 4, -1, 2, 1, -5, 4] # Example array from problem description
max_sum, max_subarray = find_max_subarray_sum(nums)
print(f"Maximum subarray sum: {max_sum}")
print(f"Subarray with maximum sum: {max_subarray}")
```

```python
def max_sum_subarray(nums):
  n=len(nums)
  maxi=float('-inf')
  for i in range(0, n):
    total=0
    for j in range(i, n):
      total=total + nums[j]
      maxi=max(maxi, total)
  return maxi

nums = [-2, 1, -3, 4, -1, 2, 1, -5, 4]
max_sum=max_sum_subarray(nums)
print(f'Max subarray sum:{max_sum}')
```

```python
# Two sum problem

def two_sum(nums, target):
  n=len(nums)
  for i in range(0, n-1):
    for j in range(i+1, n):
      if nums[i] + nums[j] == target:
        return [i, j]
  return [] # This return statement was moved outside the loops to correct logic

nums=[1, 3, 5, 9, 7, 4, 3, 2]
target=8
result = two_sum(nums, target)
print('indices:', result)
```

### Kadane's Algorithm for Maximum Subarray Sum (and the Subarray)

Kadane's algorithm is a dynamic programming approach that efficiently finds the maximum sum of a contiguous subarray. It iterates through the array, keeping track of the current maximum sum ending at the current position and the overall maximum sum found so far.

```python
# Trapping Rain Water

def trap_water(height):
  left, right = 0, len(height) - 1
  left_max = right_max = 0
  water = 0

  while left < right:
    if height[left] < height[right]:
      if height[left] >= left_max:
        left_max = height[left]
      else:
        water += left_max - height[left]
      left += 1
    else:
      if height[right] >= right_max:
        right_max = height[right]
      else:
        water += right_max - height[right]
      right -= 1
  return water

height = [0, 1, 0, 2, 1, 0, 1, 3, 2, 1, 2, 1]
trapped_water = trap_water(height)
print(f"Trapped water: {trapped_water}")
```

```python
# Find the first and last occurrence of a element
def find_first_last(nums, target):
  n=len(nums)
  first=-1
  last= -1
  for i in range(0, n):
    if nums[i]== target:
      if first == -1:
        first = i
      last = i
  return[first, last]

nums=[1, 2, 3, 3, 3, 3, 3, 5, 6, 8, 9, 9, 10]
target=3
print(find_first_last(nums, target))
```

```python
from typing import List
class Solution:
  def removeDuplicates(self, nums:List[int]) -> int:
    my_dict=dict()
    for i in nums:
      my_dict[i]=0
    j=0
    for n in my_dict:
      nums[j]=n
      j+=1
    return j

nums = [1, 1, 2 ,4 ,3, 2, 4 ,3 , 1, 5]
sol = Solution()
length =sol.removeDuplicates(nums)
print('Array after removing duplicates:', nums[:length])
print('New length:', length)
```
