```python
def factorial(num):
    if num==0 or num==1:
        return 1
    return num * factorial(num-1)


fact=factorial(4)
print('factorial of num:', fact)
```

Two sum problem

```python
# Brute Method
def max_subarray_sum(nums):
    n=len(nums)
    maxi=float('-inf')
    for i in range(0, n):
        total=0
        for j in range(i, n):
            total=total + nums[j]
            maxi= max(maxi, total)
    return maxi

nums=[-2, 1, -3, 4, -1, 2, 1, -5, 4]
print(max_subarray_sum(nums))
```

```python
# Optimal Method
def max_subarray_sum(nums):
    n=len(nums)
    maxi=float('-inf')
    total=0
    for i in range(0, n):
        total=total + nums[i]
        maxi= max(maxi, total)
        if total <0:
            total=0
    return maxi

nums=[-2, 1, -3, 4, -1, 2, 1, -5, 4]
print(max_subarray_sum(nums))
```

 Largest Elemetnt in an array

```python
def largest_ele(nums):
    largest=nums[0]
    n= len(nums)
    for i in range(0, n):
        largest=max(largest, nums[i])
    return largest

nums=[55, 32, -97, 99, 3, 67]
print('largest ele in  array:', largest_ele(nums))
# TC - O(n)
#sc - o(n)
```

```python
# 2 largest element
def sec_largest(arr):
    largest=float('-inf')
    s_largest= float('-inf')
    n=len(arr)
    for i in range(0, n):
        largest=max(largest, arr[i])
    for i in range(0, n):
        if arr[i]> s_largest and arr[i] != largest:
            s_largest= arr[i]
    return s_largest

arr=[55, 32, -97, 99, 3, 67]
print(sec_largest(arr))
```

Time to buy and sell stock

```python
#Brute Force method
def stock(prices):
    n= len(prices)     # TC - O(N(N+1)/2)= O(N2)
    max_profit = 0     #SC - O(1)
    for i in range(0, n):
        for j in range(i+1, n):
            if prices[j] > prices[i]:
                p=prices[j]- prices[i]
                max_profit=max(max_profit, p)
    return max_profit

prices=[7, 2, 3, 5, 6, 4, 9]
buy=stock(prices)
print(buy)

```

```python
#Optimal Solution
def stock(prices):
    max_profit=0
    min_price=float('inf')
    n=len(prices)
    for i in range(0, n):
        min_price=min(min_price, prices[i])
        max_profit=max(max_profit, prices[i]-min_price)
    return max_profit

prices=[7, 2, 3, 5, 11, 4, 9]
buy=stock(prices)
print(buy)
```

```python
def ele_arrange(nums):
    pos=[5, 10, 6]
    neg= [-3, -1, -10]
    for i in range (0, len(pos)):
        nums[2*i]=pos[i]
        nums[(2*i)+1]=neg[i]
    return nums

nums=[5, 10, -3, -1, -10, 6]
buy=ele_arrange(nums)
print(buy)
```

```python
class Queue:
    def __init__(self):
        self.items = []

    def is_empty(self):
        return len(self.items) == 0

    def enqueue(self, items):
        self.items.append(items)

    def dequeue(self):
        if len(self.items) == 0:
            print('dequeue from empty queue')
            return
        x = self.items.pop(0)
        return x

    def front(self):
        if len(self.items) == 0:
            print('cannot peek, queue is empty ')
            return
        return self.items[0]

    def rear(self):
        if len(self.items) == 0:
            print('cannot read , queue is empyt')
            return
        return self.items[-1]

    def size(self):
        return len(self.items)

queue=Queue()
queue.enqueue(10)
queue.enqueue(12)
queue.enqueue(19)
queue.enqueue(13)
queue.enqueue(15)
print(f'queue content= {queue.items}')
print(f'Dequue item = {queue.dequeue()}')
print(f'queue content= {queue.items}')
print(f'front item after dequeue= {queue.front()}')
print(f'queue is empty= { queue.is_empty()}')

```
