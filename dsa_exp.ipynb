```python
def merge_sort(unsorted_list):
    if len(unsorted_list)<=1:
        return unsorted_list

#find middle point and divide it
    middle=len(unsorted_list)//2
    left=unsorted_list[:middle]
    right=unsorted_list[middle:]
    left=merge_sort(left)
    right=merge_sort(right)
    return list(merge(left, right))

#merge sorted halves
def merge(left_half, right_half):
    res=[]
    while len(left_half)!=0 and len(right_half) !=0:
        if left_half[0]<right_half[0]:
            res.append(left_half[0])
            left_half.remove(left_half[0])
        else:
            res.append(right_half[0])
            right_half.remove(right_half[0])
    if len(left_half)==0:
        res=res+right_half
    else:
        res=res+left_half
    return res
unsorted_list=[64, 55, 24, 34, 34, 32, 22]
print(merge_sort(unsorted_list))
```

```python
from typing import List
from math import sqrt
num=36
def factors (num: int) -> List[int]:
    factors =[]
    for i in range(1, int(sqrt(num))+1):
        if num % i == 0:
            factors.append(i)
            if num // i != i:
                factors.append(num// i)
    factors.sort()
    return factors
print(factors(num))
```

```python
class Solution:
    def isPalindrome(self, s: str) -> bool:
        n=len(s)
        left=0
        right=n-1
        while left< right:
            if s[left] != s[right]:
                return False
            left+=1
            right-=1
        return True

s="ANBCDDCBNA"
sol= Solution()
print(sol.isPalindrome(s))
```

```python
class solution:
    def pal(self, s:str) -> bool:
        n=len(s)
        left=0
        right=n-1
        while left< right:
            if s[left] != s[right]:
                return False
            left +=1
            right -=1
        return True

s="ANBCDDCBNA"
sol = solution()
print(sol.pal(s))
```

```python
def get_ans(n, par, a):
    MOD = 10**9 + 7

    # build tree
    tree = [[] for _ in range(n)]
    for i in range(1, n):
        tree[par[i] - 1].append(i)

    # square-free transform
    def square_free(x):
        p = 2
        res = 1
        while p * p <= x:
            cnt = 0
            while x % p == 0:
                x //= p
                cnt ^= 1
            if cnt:
                res *= p
            p += 1
        if x > 1:
            res *= x
        return res

    sf = [square_free(x) for x in a]
    ans = 0

    def dfs(u):
        nonlocal ans
        mp = {}
        # include self
        mp[sf[u]] = 1
        cur_beauty = 0

        for v in tree[u]:
            child_mp, child_pairs = dfs(v)
            cur_beauty += child_pairs

            # small-to-large merge
            if len(child_mp) > len(mp):
                mp, child_mp = child_mp, mp

            for key, cnt in child_mp.items():
                if key in mp:
                    cur_beauty += mp[key] * cnt
                    mp[key] += cnt
                else:
                    mp[key] = cnt

        ans = (ans + cur_beauty) % MOD
        return mp, cur_beauty

    dfs(0)
    return ans
```
